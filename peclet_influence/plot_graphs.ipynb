```python
# Modules
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import scienceplots

plt.style.use('science')
```

```python
fig, ax = plt.subplots(figsize=(6, 4), dpi=600)
ax.set_xlabel("Time")
ax.set_ylabel("Temperature")
ax.set_title("Evolution of T1 in time")

data = np.loadtxt("Pe=10/enthalpy_method/output/LEVEL_8/data.vof")
time = data[:,0]
T1 = data[:,1]
ax.plot(time, T1, color = "blue", label = f"Pe = 100")

data = np.loadtxt("Pe=100/Re=100/enthalpy_method/output/LEVEL_8/data.vof")
time = data[:,0]
T1 = data[:,1]
ax.plot(time, T1, color = "mediumorchid", label = f"Pe = 1000")

data = np.loadtxt("Pe=1000/enthalpy_method/output/LEVEL_8/data.vof")
time = data[:,0]
T1 = data[:,1]
ax.plot(time, T1, color = "red", label = f"Pe = 10000")

plt.axhline(y=0.01407, color='k', linestyle='--', label = "Equilibrium temperature")
ax.legend()
plt.tight_layout()
plt.savefig("Figures/Evolution of T1 in time.png",dpi=600)
plt.show()
```
