```python
# Modules
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import scienceplots

plt.style.use('science')
cmap = mcolors.LinearSegmentedColormap.from_list("", ["blue","blueviolet", "mediumorchid","deeppink","red"])
norm = mcolors.Normalize(vmin=4, vmax=8)
```

```python
# We plot the graph of  evolution of enthalpy in time
fig, ax = plt.subplots(figsize=(6, 4), dpi=600)
ax.set_xlabel("Time")
ax.set_ylabel("Temperature")
ax.set_title("Evolution of temperature in time according to the method")

for i in range(3):
    if i == 0:
        data = np.loadtxt(f"enthalpy_method/output/LEVEL_8/data.vof")
        time = data[:,0]
        T1 = data[:,1]
        ax.plot(time, T1, color = "blue", label = f"Measuring the enthalpy")
    elif i == 1:
        data = np.loadtxt(f"temp_diff_method/output/LEVEL_8/data.vof")
        time = data[:,0]
        T1 = data[:,1]
        ax.plot(time, T1, color = "mediumorchid", label = f"Measuring the system's temperature")
    else:
        data = np.loadtxt(f"temperature_method/output/LEVEL_8/data.vof")
        time = data[:,0]
        T1 = data[:,1]
        ax.plot(time, T1, color = "red", label = f"Measuring both temperatures")

plt.axhline(y=0.01407, color='k', linestyle='--', label = "Equilibrium temperature")
ax.legend()
plt.tight_layout()
plt.show()
```
