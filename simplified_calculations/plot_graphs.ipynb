```python
# Modules
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import scienceplots

plt.style.use('science')
cmap = mcolors.LinearSegmentedColormap.from_list("", ["darksalmon", "coral","orangered","red","mediumslateblue","mediumpurple","blueviolet","blue","chartreuse", "yellowgreen","olivedrab","green"])
cmap2 = mcolors.LinearSegmentedColormap.from_list("", ["red","blue","green"])
norm = mcolors.Normalize(vmin=0, vmax=11)
```

```python
data_vof = np.loadtxt("/home/bilal/basilisk/LEVEL_convergence/temp_measure/without_speed/enthalpy_method/output/LEVEL_9/data.vof")
time_vof = data_vof[:,0]
T1_vof = data_vof[:,1]
erreur_T1_ent = [[] for _ in range(4)]

for i in range(4,8):
    for j in range(i+1,9):
        data_j = np.loadtxt(f"enthalpy_method/output/deltax={2**i}/LEVEL_{j}/data.vof")
        t_j = data_j[:, 0]
        T1_j = data_j[:, 1]
        T1_j_interp = np.interp(time_vof, t_j, T1_j)
        erreur_T1_j = np.abs(T1_j_interp - T1_vof)
        erreur_T1_ent[j-5].append(np.mean(np.abs(erreur_T1_j)))

data_vof = np.loadtxt("/home/bilal/basilisk/LEVEL_convergence/temp_measure/without_speed/temp_diff_method/output/LEVEL_9/data.vof")
time_vof = data_vof[:,0]
T1_vof = data_vof[:,1]
erreur_T1_td = [[] for _ in range(4)]

for i in range(4,8):
    for j in range(i+1,9):
        data_j = np.loadtxt(f"temp_diff_method/output/deltax={2**i}/LEVEL_{j}/data.vof")
        t_j = data_j[:, 0]
        T1_j = data_j[:, 1]
        T1_j_interp = np.interp(time_vof, t_j, T1_j)
        erreur_T1_j = np.abs(T1_j_interp - T1_vof)
        erreur_T1_td[j-5].append(np.mean(np.abs(erreur_T1_j)))

data_vof = np.loadtxt("/home/bilal/basilisk/LEVEL_convergence/temp_measure/without_speed/temperature_method/output/LEVEL_9/data.vof")
time_vof = data_vof[:,0]
T1_vof = data_vof[:,1]
erreur_T1_temp = [[] for _ in range(4)]

for i in range(4,8):
    for j in range(i+1,9):
        data_j = np.loadtxt(f"temperature_method/output/deltax={2**i}/LEVEL_{j}/data.vof")
        t_j = data_j[:, 0]
        T1_j = data_j[:, 1]
        T1_j_interp = np.interp(time_vof, t_j, T1_j)
        erreur_T1_j = np.abs(T1_j_interp - T1_vof)
        erreur_T1_temp[j-5].append(np.mean(np.abs(erreur_T1_j)))
```

```python
delta1 = np.array([2**(-4),2**(-5),2**(-6),2**(-7)])
err = [erreur_T1_ent,erreur_T1_td,erreur_T1_temp]
lab = ["LEVEL = 5, Method 1","LEVEL = 6, Method 1","LEVEL = 7, Method 1","LEVEL = 8, Method 1","LEVEL = 5, Method 2","LEVEL = 6, Method 2","LEVEL = 7, Method 2","LEVEL = 8, Method 2","LEVEL = 5, Method 3","LEVEL = 6, Method 3","LEVEL = 7, Method 3","LEVEL = 8, Method 3"]
sym = ['o','*','+']
power1 = []

fig, ax = plt.subplots(figsize=(7, 4), dpi=600)
ax.set_xlabel(r"$\Delta_1$")
ax.set_ylabel("Mean error")
ax.set_yscale("log")
ax.set_xscale("log")
ax.set_title(r"Mean error for T1 depending on $\Delta_1$")

for i in range(3):
    ax.plot(delta1[:1],err[i][0], sym[i], color = cmap(norm(4*i)), label=lab[4*i])
    ax.plot(delta1[:2],err[i][1], sym[i], color = cmap(norm(4*i+1)), label=lab[4*i+1])
    ax.plot(delta1[:3],err[i][2], sym[i], color = cmap(norm(4*i+2)), label=lab[4*i+2])
    ax.plot(delta1[:4],err[i][3], sym[i], color = cmap(norm(4*i+3)), label=lab[4*i+3])
    power1.append(np.polyfit(np.log(delta1[:4]),np.log(err[i][3]),deg=1)[0])

print(power1)
ax.legend(bbox_to_anchor=(1, 1), loc='upper left', fontsize='small')
plt.tight_layout()
plt.savefig("Figures/Mean error for T1.png")
plt.show()
```
